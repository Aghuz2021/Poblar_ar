```python
import pandas as pd
import os
archivo = os.path.join("..", "Data", "Bd.csv")
df = pd.read_csv(archivo)
df["Total habitantes"] = pd.to_numeric(
            df["Total habitantes"].astype(str)
            .str.replace('.', '', regex=False)
            .str.replace(',', '', regex=False), 
            errors='coerce'
        )
df["Varon/Masculino"] = pd.to_numeric(
            df["Varon/Masculino"].astype(str)
            .str.replace('.', '', regex=False)
            .str.replace(',', '', regex=False), 
            errors='coerce'
        )
df
```

```python


import pandas as pd
from tabulate import tabulate  # Asegúrate de tenerlo instalado: pip install tabulate

def varonesProvincia(df):
    """
    Filtra por Provincia, agrupa por Nombre, suma los varones,
    y muestra la tabla completa con un formato visual ordenado.
    """
    try:
        resultado = (
            df.query("`Tipo de jurisdicción` == 'Provincia'")
            .groupby("Nombre")["Varon/Masculino"]
            .sum()
            .reset_index()
            .sort_values(by="Varon/Masculino", ascending=False)
        )
        resultado["Varon/Masculino"] = resultado["Varon/Masculino"].apply(
            lambda x: f"{int(x):,}".replace(",", ".") if pd.notnull(x) else ""
        )
        # Imprimir usando tabulate con un diseño estético de grilla y sin mostrar el índice numérico
        print(tabulate(
            resultado, 
            headers=['Provincia', 'Total Varones'], 
            tablefmt='fancy_grid', 
            showindex=False
        ))
        
    except Exception as e:
        print(f"[Error en varonesProvincia]: {e}")

# Llamada a la función
varonesProvincia(df)
```

```python
import pandas as pd
from tabulate import tabulate  # pip install tabulate

def count_varones_argentina(df):
    """
    Filtra por Provincia y calcula la mediana total de habitantes 
    de todo el conjunto, sin agrupar por nombre.
    """
    try:
        # 1. Filtrar por Provincia y calcular la mediana directa de la columna
        mediana_valor = (
           df.query("`Tipo de jurisdicción` in ['Provincia', 'Jurisdicción autónoma']")["Varon/Masculino"]
            .sum()
        )
        
        # 2. Formatear el número con punto como separador de miles (ej: 300.000)
        mediana_formateada = f"{int(mediana_valor):,}".replace(",", ".") if pd.notnull(mediana_valor) else ""
        
        # 3. Crear una tabla de una sola fila para mostrar el resultado con tabulate
        resultado = pd.DataFrame({
            "Indicador": ["Total Varones"],
            "Total Varones": [mediana_formateada]
        })
        
        # 4. Imprimir con diseño estético
        print(tabulate(
            resultado,  # type: ignore
            headers='keys', 
            tablefmt='fancy_grid', 
            showindex=False
        ))
        
    except Exception as e:
        print(f"[Error en count_varones_argentina]: {e}")

# Llamada a la función
count_varones_argentina(df)
```
